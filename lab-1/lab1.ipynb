# Tasks

```
import pandas as pd
from pathlib import Path

CSV_PATH = Path("assignment_materials_repository.csv")
OUT_PATH = Path("student_submission_lab1.csv")


print("=== Task 8.1 ===")
df_eval = pd.read_csv(CSV_PATH)
df_battery = df_eval.copy()

# energy_density_proxy = |ΔEf| / ρ²
df_battery["energy_density_proxy"] = (
    df_battery["formation_energy"].abs() / df_battery["density"] ** 2
).round(4)

# Top-5 materials by highest energy_density_proxy
top_5 = (
    df_battery[["formula", "density", "formation_energy", "energy_density_proxy"]]
    .sort_values("energy_density_proxy", ascending=False)
    .head(5)
)
print(top_5.to_string(index=False))


print("\n\n=== Task 8.2 ===")

mask = (
    (df_eval["band_gap"] >= 0.1)
    & (df_eval["band_gap"] <= 1.6)
    & (df_eval["density"] >= 5.0)
)

thermoelectric_candidates = df_eval[mask].copy()

print(f"Candidates passing both criteria: {len(thermoelectric_candidates)}\n")
print(
    thermoelectric_candidates[["material_id", "formula", "band_gap", "density"]]
    .sort_values("density", ascending=False)
    .to_string(index=False)
)


print("\n\n=== Task 8.3 ===")

missing_counts = df_eval.isna().sum()
missing_pct = (missing_counts / len(df_eval) * 100).map("{:.2f}".format)
audit = pd.DataFrame({"missing_count": missing_counts, "missing_pct": missing_pct})
print("Missing-value audit:")
print(audit.to_string())

metal_nan_mask = (df_eval["is_metal"] == True) & (df_eval["band_gap"].isna())
df_eval.loc[metal_nan_mask, "band_gap"] = 0.0
imputed_count = metal_nan_mask.sum()
print(f"\nImputed band_gap = 0.0 for {imputed_count} confirmed-metal row(s).")
print(f"Remaining NaN in band_gap: {df_eval['band_gap'].isna().sum()}")

df_eval = df_eval.dropna(subset=["band_gap"])
print(f"Rows after pruning non-metal NaN band_gap: {len(df_eval)}")

df_eval = df_eval.sort_values("formation_energy", ascending=True).reset_index(drop=True)

df_eval.to_csv(OUT_PATH, index=False)
print(f"\nCleaned dataset exported to '{OUT_PATH}' ({len(df_eval)} rows).")
```
